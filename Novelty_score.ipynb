### Step 3: Novelty Assessment using TF-IDF and Cosine Similarity

We will use TF-IDF to convert the text of the new grant and the simulated research abstracts into numerical representations. Then, we'll calculate the cosine similarity between the grant and each abstract. Lower similarity scores generally indicate higher novelty.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Combine the new grant text with the simulated abstracts for TF-IDF training
all_documents = [new_grant_text] + simulated_research_abstracts

# Initialize TF-IDF Vectorizer
# min_df ignores terms that appear in too few documents
# stop_words removes common words like 'the', 'is', 'a', etc.
tfidf_vectorizer = TfidfVectorizer(min_df=1, stop_words='english')

# Fit the vectorizer to all documents and transform them
tfidf_matrix = tfidf_vectorizer.fit_transform(all_documents)

# The first vector is for the new grant, the rest are for the research abstracts
grant_vector = tfidf_matrix[0:1]
research_vectors = tfidf_matrix[1:]

# Calculate cosine similarity between the new grant and each research abstract
similarity_scores = cosine_similarity(grant_vector, research_vectors).flatten()

print("Cosine Similarity Scores (New Grant vs. Research Abstracts):")
for i, score in enumerate(similarity_scores):
    print(f"  Research Abstract {i+1}: {score:.4f}")

# Let's also identify the most and least similar abstracts
most_similar_index = similarity_scores.argmax()
least_similar_index = similarity_scores.argmin()

print(f"\nMost similar abstract (Score: {similarity_scores[most_similar_index]:.4f}):")
print(f"  '{simulated_research_abstracts[most_similar_index][:150]}...'\n")

print(f"Least similar abstract (Score: {similarity_scores[least_similar_index]:.4f}):")
print(f"  '{simulated_research_abstracts[least_similar_index][:150]}...'\n")

print("\nIn the next step, we'll use these scores to calculate an overall novelty score for the grant.")
```

### Function for Novelty Assessment

To easily compare multiple grant drafts, let's encapsulate the entire novelty assessment process into a function. This function will take a `grant_text` and the `simulated_research_abstracts` as input, and return the calculated novelty score and an explanation.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import yake

def assess_grant_novelty(grant_text: str, research_abstracts: list, top_k_keywords: int = 5):
    """
    Assesses the novelty of a grant text against a list of existing research abstracts.

    Args:
        grant_text (str): The text of the new grant proposal.
        research_abstracts (list): A list of strings, where each string is an abstract
                                   of an existing research paper.
        top_k_keywords (int): The number of top keywords to extract for context.

    Returns:
        tuple: A tuple containing (novelty_score, novelty_explanation_text).
               novelty_score is a float between 0 and 1.
               novelty_explanation_text is a string explaining the novelty.
    """

    # 1. Keyword Extraction
    kw_extractor = yake.KeywordExtractor(lan="en", n=2, dedupLim=0.9, top=20, features=None)
    keywords = kw_extractor.extract_keywords(grant_text)
    top_keywords = [kw for kw, score in keywords[:top_k_keywords]]

    # 2. TF-IDF Vectorization and Cosine Similarity
    all_documents = [grant_text] + research_abstracts
    tfidf_vectorizer = TfidfVectorizer(min_df=1, stop_words='english')
    tfidf_matrix = tfidf_vectorizer.fit_transform(all_documents)

    grant_vector = tfidf_matrix[0:1]
    research_vectors = tfidf_matrix[1:]

    similarity_scores = cosine_similarity(grant_vector, research_vectors).flatten()

    # 3. Calculate Novelty Score
    max_similarity = np.max(similarity_scores)
    final_novelty_score = 1 - max_similarity # Higher score means more novel

    # 4. Generate Explanation
    novelty_explanation = []

    if final_novelty_score > 0.75:
        novelty_explanation.append("The grant exhibits *very high potential for novelty*.")
    elif final_novelty_score > 0.5:
        novelty_explanation.append("The grant shows *good potential for novelty*.")
    elif final_novelty_score > 0.25:
        novelty_explanation.append("The grant has *moderate novelty*.")
    else:
        novelty_explanation.append("The grant appears to have *limited novelty*.")

    novelty_explanation.append(f"\n**Reasoning for Novelty:**\n- Calculated Novelty Score (0-1): {final_novelty_score:.4f}")
    novelty_explanation.append(f"- Highest Similarity with any existing research: {max_similarity:.4f}")

    most_similar_index = similarity_scores.argmax()
    least_similar_index = similarity_scores.argmin()

    novelty_explanation.append(f"- Most similar abstract (Index {most_similar_index+1}, Score: {max_similarity:.4f}): '{research_abstracts[most_similar_index][:100]}...' ")
    novelty_explanation.append(f"- Least similar abstract (Index {least_similar_index+1}, Score: {similarity_scores[least_similar_index]:.4f}): '{research_abstracts[least_similar_index][:100]}...' ")

    if top_keywords:
        novelty_explanation.append(f"- Key themes from grant: {', '.join(top_keywords)}")

    return final_novelty_score, " ".join(novelty_explanation)

print("Novelty assessment function `assess_grant_novelty` defined.")
```
